```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.arange(1, 101)
y1 = np.loadtxt('dice.txt')

x2 = x1
y2 = np.loadtxt('penalty_dice.txt')

x3 = x1
y3 = np.loadtxt('bonus_dice.txt')

x4 = np.arange(1, 19)
y4 = np.loadtxt('three_dice_pool.txt')

x5 = np.arange(1, 201)
y5 = np.loadtxt('double_dice.txt')
```

```python
plt.figure(figsize=(10, 6))


plt.bar(x1, y1, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Dice, 100-гранный кубик')
plt.xlabel('Значение на кубике')
plt.ylabel('Вероятность выпадения')
plt.grid(alpha=0.3)


plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(x2, y2, alpha=0.7, color='lightcoral', edgecolor='black')
plt.title('Penalty Dice, 100-гранный кубик')
plt.xlabel('Значение на кубике')
plt.ylabel('Вероятность выпадения')
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(x3, y3, alpha=0.7, color='plum', edgecolor='black')
plt.title('Bonus Dice, 100-гранный кубик')
plt.xlabel('Значение на кубике')
plt.ylabel('Вероятность выпадения')
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(x4, y4, alpha=0.7, color='lightgreen', edgecolor='black')
plt.title('Three Dice Pool, 6-гранный кубик')
plt.xlabel('Сумма значений на 3-х кубиках')
plt.ylabel('Вероятность выпадения')
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(x5, y5, alpha=0.7, color='gold', edgecolor='black')
plt.title('Double Dice Pool, 100-гранный кубик')
plt.xlabel('Сумма значений Penalty Dice и Bonus Dice')
plt.ylabel('Вероятность выпадения')
plt.grid(alpha=0.3)

plt.show()
```
