```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def k(x, y):
    mean_x = np.mean(x)
    mean_y = np.mean(y)
    mean_xx = np.mean(x*x)
    mean_yy = np.mean(y*y)
    mean_xy = np.mean(x*y)

    return (mean_xy - mean_x*mean_y)/(mean_xx - mean_x**2)
```

```python
def b(x, y):
    mean_x = np.mean(x)
    mean_y = np.mean(y)

    return mean_y - k(x, y)*mean_x
```

```python
time1 = np.loadtxt('data_1.1.txt')
N1 = np.arange(1, len(time1)+1) * 100

x1 = np.log(N1)
y1 = np.log(time1)

k1 = k(x1, y1)
b1 = b(x1, y1)
```

```python
plt.scatter(x1, y1, s=5)

plt.plot(x1, k1*x1 + b1, label=f'k1 = {k1}', c='orange')

plt.xlabel('ln(N)')
plt.ylabel('ln(t)')

plt.legend()
plt.title(1.1)
plt.grid(True)
plt.show()
```
