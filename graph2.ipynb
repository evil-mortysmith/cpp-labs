```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def k(x, y):
    mean_x = np.mean(x)
    mean_y = np.mean(y)
    mean_xx = np.mean(x*x)
    mean_yy = np.mean(y*y)
    mean_xy = np.mean(x*y)

    return (mean_xy - mean_x*mean_y)/(mean_xx - mean_x**2)
```

```python
def b(x, y):
    mean_x = np.mean(x)
    mean_y = np.mean(y)

    return mean_y - k(x, y)*mean_x
```

```python
push_times = np.loadtxt('push_times.txt')
poll_times = np.loadtxt('poll_times.txt')
n = np.arange(100, 100 + len(push_times))
```

```python
x1 = np.log(n)
y1 = np.log(push_times)

k1 = k(x1, y1)
b1 = b(x1, y1)
```

```python
x2 = np.log(n)
y2 = np.log(poll_times)

k2 = k(x2, y2)
b2 = b(x2, y2)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# График для queue_push
ax1.scatter(x1, y1, s=10, color='blue', alpha=0.5)
ax1.plot(x1, k1*x1 + b1, label=f'k1 = {k1}', c='orange')
ax1.set_xlabel('ln(N)')
ax1.set_ylabel('t, мкс')
ax1.set_title('2.1')
ax1.grid(True)
ax1.legend()

# График для queue_poll
ax2.scatter(x2, y2, s=10, color='blue', alpha=0.5)
ax2.plot(x2, k2*x2 + b2, label=f'k2 = {k2}', c='orange')
ax2.set_xlabel('ln(N)')
ax2.set_ylabel('t, мкс')
ax2.set_title('2.2')
ax2.grid(True)
ax2.legend()


plt.show()
```
