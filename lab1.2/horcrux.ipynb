```python
import matplotlib.pyplot as plt
import numpy as np

N = np.loadtxt('arr_length.txt')
search_time = np.loadtxt('search_time.txt')
fast_search_time = np.loadtxt('fast_search_time.txt')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 8))
axes[0].scatter(N, search_time, label='Search', marker='o', s = 5)

axes[1].scatter(N, fast_search_time, label='Fast Search', marker='o', s = 5)

plt.show()
```
