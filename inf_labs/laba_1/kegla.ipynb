```python
import matplotlib.pyplot as plt
import numpy as np

N = np.loadtxt('arr_length.txt')
linear_time = np.loadtxt('linear_search_time.txt')  # Время для линейного поиска
binary_time = np.loadtxt('binary_search_time.txt')  # Время для бинарного поиска
rand_linear_time = np.loadtxt('rand_linear_search_time.txt')  # Время для случайного линейного поиска
rand_binary_time = np.loadtxt('rand_binary_search_time.txt')  # Время для случайного бинарного поиска
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 8))
axes[0, 0].scatter(N, linear_time, label='Linear Search', marker='o', s = 5)
axes[0, 0].set_title('Time complexity of the linear search')

axes[0, 1].scatter(N, binary_time, label='Binary Search', marker='o', s = 5)
axes[0, 1].set_title('Time complexity of the binary search')

axes[1, 0].scatter(N, rand_linear_time, label='Random Linear Search', marker='o', s = 5)
axes[1, 0].set_title('Time complexity of the random linear search')

axes[1, 1].scatter(N, rand_binary_time, label='Random Binary Search', marker='o', s = 5)
axes[1, 1].set_title('Time complexity of the random binary search')

#plt.xlabel('Number of elements (N)')
#plt.ylabel('Time (ms)')
#plt.title('Time complexity of search algorithms')
#plt.legend()
#plt.grid(True)
plt.show()
```
